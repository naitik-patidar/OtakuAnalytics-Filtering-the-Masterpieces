```python
import pandas as pd
import numpy as np
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("CooperUnion/anime-recommendations-database")

```

```python
df = pd.read_csv(path + "/anime.csv")
```

```python
df
```

```python
df.head()
```

```python
df.loc[5] ["rating"]
```

```python
def rating_count(rating):
    if rating >= 9:
        return "Excellent"
    if rating >= 8:
        return "Good"
    if rating >= 7:
        return "Average"
    else:
        return "Poor"
```

```python
df["rating_category"] = df["rating"].apply(rating_count)
```

```python
df
```

```python
excellent_anime = df[df["rating_category"] == "Excellent"]
excellent_anime
```

```python
poor_anime = df[df["rating_category"] == "Poor"]
poor_anime
```

```python
avg_anime = df[df['rating_category'] == "Average"]
avg_anime
```

```python
df
```

```python
rating_mean = df["rating"].mean()
```

```python
rating_mean
```

```python
best_to_watch_rating_wise = df[df["rating"] > rating_mean]
best_to_watch_rating_wise
```

```python
members_mean = df["members"].mean()
members_mean
```

```python
best_to_watch_members_rating_mean_wise = df[(df["rating"] > rating_mean) & (df["members"] > members_mean)]
best_to_watch_members_rating_mean_wise
```

```python
# This groups the data by type and calculates the mean rating for each
df.groupby("type")["rating"].mean()
```

```python
top_anime = df.sort_values(by ="rating", ascending = False).head(10) 
top_anime
```

```python
popular_anime = df.sort_values(by = "members" , ascending = False).head(100)

top_famous = popular_anime.sort_values(by= "rating", ascending = False).head(100)

top_famous
```

```python
print(top_famous[['name', 'genre', 'type', 'episodes', 'rating', 'members']].head(100).to_string())
```
